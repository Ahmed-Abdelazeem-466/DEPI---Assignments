## 8. OOP — Solution

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    def info(self):
        return f"{self.name} has grade {self.grade}"

s = Student('Omar', 90)
s.info()
```

## 9. Inheritance — Solution

```python
class GraduateStudent(Student):
    def thesis_topic(self):
        return 'Generative AI Optimization'

gs = GraduateStudent('Laila', 95)
print(gs.info())
print(gs.thesis_topic())
```

## 10. Decorators — Solution

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print('Running function...')
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(a, b):
    return a * b

multiply(4, 3)
```

## 11. Generators — Solution

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

list(countdown(5))
```

## 12. Lambda, Map, Filter, Reduce — Solution

```python
from functools import reduce

numbers = [1,2,3,4,5]
squares = list(map(lambda x: x*x, numbers))
odds = list(filter(lambda x: x%2!=0, numbers))
total = reduce(lambda a,b: a+b, numbers)
squares, odds, total
```

## 13. Advanced Practice — Solutions

```python
def analyze_list(lst):
    return max(lst), min(lst), sum(lst)/len(lst), [x for x in lst if x%2==0]

analyze_list([1,2,3,4,5,6])
```

```python
class Agent:
    def __init__(self, name, role):
        self.name = name
        self.role = role
    def act(self):
        print(f"Agent {self.name} is performing {self.role} action.")

a = Agent('Alpha','Navigation')
a.act()
```

```python
def divisible_by_3():
    for i in range(51):
        if i % 3 == 0:
            yield i

list(divisible_by_3())
```
